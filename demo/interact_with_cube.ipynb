## (B) Let's now solve using `cross_solver()`
- the solver will go through each different type of solver above and apply different solutions; it will search through different options using a tree method
- it will consider the different types after each sub-attempt, as the types will be dynamic after each attempt since the cube state is constantly moving after every solved sub-step

```python
from cube import cross_solver
solutions = cross_solver(_cube)
```

```python
solutions
```

## (C) notice that we may compress one of the solutions actually since Bp Bp --> B2 (2 moves --> 1 move)

```python
from cube import compress_moves
```

```python
local_min = float("inf")
final_solutions = []

# quick way to take compressed solution if indeed there is a solution that reduces in len()
for solution in solutions:
    compressed_solution = compress_moves(solution[1])
    L = len(compressed_solution)
    if L < local_min:      
        final_solutions.append(compressed_solution)
        local_min = L  

solution = final_solutions[-1]
print(len(solution), solution)
```

```python
solution_converted = move_notation_converter(solution)
solution_converted
```

```python
scramble_converted = move_notation_converter(scramble)
scramble_converted
```

```python
scramble_then_solution = scramble+solution
print(scramble_then_solution)
```
