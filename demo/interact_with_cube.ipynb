# Important-Considerations
- Recall that a `cube` is represented as follows for a given face, when looking at it head-on with yellow on top, and/or green front
    > [0,1,2]

    > [7,8,3]

    > [6,5,4] 
- Furthermore, all moves are assuming ***green front, yellow top***

# Scrambling a `cube`
- one may use `do_scramble()` with a `list` of `strings` as moves such as:
    - `do_scramble(["R", "U", "R"])`
- the below is a `cube` object since `do_scramble` performs moves on `solved_cube()` as a default
    - this can be changed and will be explored further


```python
# this is now a cube object, 
cube = do_scramble(["R"])
```

```python
cube.cube_state
```

## More applications of `do_scramble()`
- `do_scramble` actually can apply moves with a starting position of a `solved_cube()`, or with a `cube` that already has moves done on it (see below)

```python
help(do_scramble)
```

```python
cube_sample1 = do_scramble(["R", "U"])
cube_sample1.cube_state
```

```python
cube_undo_sample1 = do_scramble(["Up", "Rp"], cube=cube_sample1)
cube_undo_sample1.cube_state == solved_cube().cube_state
```

# Using multiple scrambles
- You can use many scrambles if you want to explore mulitple scrambles on different cubes if need be by utilizing `iterate_through_scrambles_for_testing()`
    - `scrambles = [[moves_i], [moves_n]]`
    - `iterate_through_scrambles_for_testing(scrambles)`


```python
scrambles = [
    [
        "R2",
        "Fp",
        "Dp",
        "B",
        "D2",
        "Lp",
        "Fp",
        "D",
        "F",
        "R2",
        "D2",
        "B2",
        "Rp",
        "F",
        "R",
    ],
    [
        "R",
        "U",
        "Rp"
    ]
]

cubes = iterate_through_scrambles_for_testing(scrambles)
cube_0 = cubes[0]

# equivalently
scramble = scrambles[0]
cube = do_scramble(scramble)
```

# Applying sub cross solvers
- you can use different cross solvers on a cube
    - the `config`_type_cross_solver() depends on which "position" the white edge is in:
        - `seven_type_cross_solver()`
        - `three_type_cross_solver()`
        - `one_type_cross_solver()`
        - `top_type_cross_solver()`
        - `five_type_cross_solver()`
        - `bottum_type_cross_solver()`
    
    
        

```python
scramble = scrambles[0]
cube = do_scramble(scramble)
```

## Outputs of these sub cross-sovlers
- Notice that the output above returns a list of lists of moves
- The solver considered the different ways to solve just the "seven_type" white edge cross-piece
- Repeated outputs are normal as it considers it in relation to all of the other white edges
    - These are simplified and reduced in the full cross solver

```python
cube.seven_type_cross_solver()
```
