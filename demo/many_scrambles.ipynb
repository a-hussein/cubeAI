```python
from cube import generate_random_scramble, \
    iterate_through_scrambles_for_testing, \
    move_notation_converter, \
    cross_solver, \
    compress_moves, \
    do_scramble

from cube.visualizer import open_in_twizzle

import random
```

```python
random.seed(103195)
```

```python
many_scrambles = []
for i in range(5):
    many_scrambles.append(generate_random_scramble(num_moves=20))
```

```python
scramble0 = many_scrambles[0] 
scramble1 = many_scrambles[1] 
scramble2 = many_scrambles[2] 
scramble3 = many_scrambles[3] 
scramble4 = many_scrambles[4] 
```

```python
move_notation0=move_notation_converter(scramble0)
move_notation1=move_notation_converter(scramble1)
move_notation2=move_notation_converter(scramble2)
move_notation3=move_notation_converter(scramble3)
move_notation4=move_notation_converter(scramble4)
```

```python
move_notations = [
    move_notation0,
    move_notation1,
    move_notation2,
    move_notation3,
    move_notation4
]
```

```python
# next step is to solve the above, and compress, and visualize before and after, and confirm all solved
for set_of_moves in move_notations:
    open_in_twizzle(set_of_moves)
```

# Let's now prepare our solutions for each scramble

```python
many_cubes = iterate_through_scrambles_for_testing(many_scrambles)
```

```python
for i, cube in enumerate(many_cubes):
    k = cross_solver(cube, min_move_only=True)
    print(f"cube{i}")
    print(k)
    print("#####################")
```

```python
final_solutions = {}
for i, cube in enumerate(many_cubes):
    k = cross_solver(cube, min_move_only=True)

    local_min = float("inf")

    for solution in k:
        compressed_solution = compress_moves(solution[1])
        L = len(compressed_solution)
        if L < local_min:          
            final_solutions[f"cube{i}"] = compressed_solution
            local_min = L  


    print(f"cube{i}")     
    print("k: ", k)
    print("\t" ,f"a compressed solution for cube{i}")
    print("\t", len(final_solutions[f"cube{i}"]), final_solutions[f"cube{i}"])
    print("#####################")
```

```python
final_solutions
```

# verify that for each scrmabled cube, the cross was not solved, and is now solved

```python
for cube in many_cubes:
    if not cube.cross_solved():
        print("cube's white cross is not solved")
    else:
        print("YES, cube's white cross is solved!")

```

```python
for i, cube in enumerate(many_cubes):
    solved_cube = do_scramble(cube=cube, moves=final_solutions[f"cube{i}"])

    if not solved_cube.cross_solved():
        print("cube's white cross is not solved")
    else:
        print("YES, cube's white cross is solved!")
```
