```python
from cube import generate_random_scramble, \
    iterate_through_scrambles_for_testing, \
    move_notation_converter, \
    cross_solver, \
    compress_moves, \
    do_scramble

from cube.visualizer import open_in_twizzle

import random
```

```python
random.seed(1)
```

```python
many_scrambles = []
for i in range(5):
    many_scrambles.append(generate_random_scramble(num_moves=20))
```

```python
scramble_0 = many_scrambles[0] 
scramble_1 = many_scrambles[1] 
scramble_2 = many_scrambles[2] 
scramble_3 = many_scrambles[3] 
scramble_4 = many_scrambles[4] 
```

```python
move_converted_0=move_notation_converter(scramble_0)
move_converted_1=move_notation_converter(scramble_1)
move_converted_2=move_notation_converter(scramble_2)
move_converted_3=move_notation_converter(scramble_3)
move_converted_4=move_notation_converter(scramble_4)
```

```python
move_notations = [
    move_converted_0,
    move_converted_1,
    move_converted_2,
    move_converted_3,
    move_converted_4
]
```

```python
for move_notation in move_notations:
    open_in_twizzle(move_notation)
```

# Let's now prepare our solutions for each scramble

```python
many_cubes = iterate_through_scrambles_for_testing(many_scrambles)
```

```python
for i, cube in enumerate(many_cubes):
    solutions = cross_solver(cube)
    print(f"cube{i}")
    print(solutions)
    print("#####################")
```

```python
# quick way to return compressed solution
final_solutions = {}
for i, cube in enumerate(many_cubes):
    k = cross_solver(cube, min_move_only=True)

    local_min = float("inf")

    for solution in k:
        compressed_solution = compress_moves(solution[1])
        L = len(compressed_solution)
        if L < local_min:          
            final_solutions[f"cube{i}"] = compressed_solution
            local_min = L  


    print(f"cube{i}")     
    print("k: ", k)
    print("\t" ,f"a compressed solution for cube{i}")
    print("\t", len(final_solutions[f"cube{i}"]), final_solutions[f"cube{i}"])
    print("#####################")
```

```python
final_solutions
```

# verify that for each scrmabled cube, the cross was not solved, and is now solved

```python
for cube in many_cubes:
    if not cube.cross_solved():
        print("cube's white cross is not solved")
    else:
        print("YES, cube's white cross is solved!")

```

```python
for i, cube in enumerate(many_cubes):
    solved_cube = do_scramble(cube=cube, moves=final_solutions[f"cube{i}"])

    if not solved_cube.cross_solved():
        print("cube's white cross is not solved")
    else:
        print("YES, cube's white cross is solved!")
```
